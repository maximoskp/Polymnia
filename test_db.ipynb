```python
import sqlite3

DB_PATH = (
    "/media/maindisk/maximos/repos/data/"
    "odigos_dioikitikon_diadikasion/vectordb/"
    "chroma.sqlite3"
)

conn = sqlite3.connect(DB_PATH)

cursor = conn.cursor()

cursor.execute("""
    SELECT name
    FROM sqlite_master
    WHERE type='table'
""")

print("SQLite tables:")

for row in cursor.fetchall():
    print(row[0])

print("\nCollection records:")

try:
    cursor.execute("""
        SELECT id, name
        FROM collections
    """)

    for row in cursor.fetchall():
        print(row)

except sqlite3.Error as e:
    print("Could not inspect collections:", e)

conn.close()
```
